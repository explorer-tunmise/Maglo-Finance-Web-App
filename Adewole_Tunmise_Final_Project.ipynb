```python
normal_df = pd.read_csv("BC-TCGA-Normal.csv")

normal_df.head()
```

```python
tumor_df = pd.read_csv("BC-TCGA-Tumor.txt", sep="\t")

tumor_df.to_csv("BC-TCGA-Tumor.csv", index=False)
tumor_df.head()
```

```python
# Set 'Hybridization REF' as index for both dataframes
normal_df_indexed = normal_df.set_index('Hybridization REF')
tumor_df_indexed = tumor_df.set_index('Hybridization REF')

# Transpose the dataframes so that samples become rows and genes become columns
normal_df_transposed = normal_df_indexed.T
tumor_df_transposed = tumor_df_indexed.T

# Remove the 'Hybridization REF' name from the columns (after transpose, it becomes the columns' name)
normal_df_transposed.columns.name = None
tumor_df_transposed.columns.name = None

# Add a 'type' column to each transposed dataframe
normal_df_transposed['type'] = 0  # Label 0 for normal samples
tumor_df_transposed['type'] = 1   # Label 1 for tumor samples

# Concatenate the two dataframes
combined_df = pd.concat([normal_df_transposed, tumor_df_transposed])

# Reset the index to numbers (default integer index) as requested by the user
combined_df = combined_df.reset_index(drop=True)

print(f"Shape of combined_df: {combined_df.shape}")
display(combined_df.head())
```

```python
combined_df
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, label_binarize, MinMaxScaler
from sklearn.feature_selection import mutual_info_classif
from sklearn.ensemble import RandomForestClassifier
```

```python
datanull = combined_df.isnull().sum()
print(datanull)
g = [i for i in datanull if i > 0]
datanull[datanull > 0]
```

1. Data splitting with "iloc"- Integer Location-Based Indexing


```python
x = combined_df.iloc[:,0:-1] #Features
y = combined_df.iloc[:,-1] #class=what you wan the project to predict
```

### Handling Missing Values

Based on the previous `isnull().sum()` check, there are still columns with missing values. We will drop these columns from our feature set `x`.

```python
# Get the list of columns with missing values (where sum of nulls > 0)
columns_to_drop = datanull[datanull > 0].index.tolist()

# Drop these columns from the features DataFrame x
x_cleaned = x.drop(columns=columns_to_drop)

x_cleaned
```

Splitting the dataset into training set and testing set

```python
# Split the data using x_cleaned (without NaN columns) for features
x_train, x_test, y_train, y_test = train_test_split(x_cleaned, y, test_size=0.2, random_state=42)

print(f"Shape of x_train after cleaning and splitting: {x_train.shape}")
print(f"Shape of x_test after cleaning and splitting: {x_test.shape}")
display(x_train.head())
```

```python
min_max_scaler = MinMaxScaler()

# Fit the scaler on the training data and transform both train and test data
x_train_normalized = min_max_scaler.fit_transform(x_train)
x_test_normalized = min_max_scaler.transform(x_test) # Use transform, not fit_transform for test set

```

Feature selection using mutual info classif


```python
MI = mutual_info_classif(x_train_normalized, y_train)
MI
```

```python
n_features = 300
selected_indices = np.argsort(MI)[::-1][0:n_features]
selected_indices
```

```python
x_train_selected = x_train_normalized[:, selected_indices]
x_test_selected = x_test_normalized[:, selected_indices]

x_train_selected
```

```python
RF = RandomForestClassifier(random_state=42)
RF.fit(x_train_selected, y_train)

y_predict = RF.predict(x_test_selected)
y_proba = RF.predict_proba(x_test_selected)
```

### Model Evaluation

To assess the performance of our `RandomForestClassifier`, we'll calculate several key metrics:

*   **Accuracy Score:** The proportion of correctly classified instances.
*   **Confusion Matrix:** A table that describes the performance of a classification model on a set of test data for which the true values are known.
*   **Classification Report:** Provides a more detailed breakdown of precision, recall, and F1-score for each class.

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Calculate Accuracy
accuracy = accuracy_score(y_test, y_predict)
print(f"Accuracy Score: {accuracy:.4f}")

# Generate Confusion Matrix
conf_matrix = confusion_matrix(y_test, y_predict)
print("\nConfusion Matrix:")
print(conf_matrix)

# Generate Classification Report
class_report = classification_report(y_test, y_predict)
print("\nClassification Report:")
print(class_report)
```
